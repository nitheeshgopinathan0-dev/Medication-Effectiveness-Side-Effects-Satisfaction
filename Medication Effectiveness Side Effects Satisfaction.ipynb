<h2 style="color:#AD1457;">
⚙️ Phase 3: Feature Engineering
</h2>

<p style="font-size:15px; line-height:1.6;">

Feature engineering was performed to enhance the dataset by creating additional analytical features from the existing patient drug review data.

<br><br>

New variables were derived to transform categorical and textual information into meaningful numerical features. These engineered features include effectiveness scores, side-effect severity scores, high-satisfaction indicators, and review word-count features.

<br><br>

These additional features helped analyze patient satisfaction, measure medication effectiveness, examine the impact of side-effect severity, study review length patterns, and generate more meaningful insights during exploratory data analysis.

</p>

</div>

<div style="background-color: #3f51b5; color: white; padding: 12px 20px; border-radius: 8px; font-size: 20px; font-weight: bold; font-family: sans-serif;">
  Feature Engineering & Dataset Dimensionality Expansion
</div>

```python
# A. Text Character Lengths & Word Counts
review_columns = ['benefitsReview', 'sideEffectsReview', 'commentsReview']

for col in review_columns:
    clean_series = df[col].fillna("")
    df[f'{col}_char_len'] = clean_series.apply(len)
    df[f'{col}_word_count'] = clean_series.apply(lambda x: len(x.split()))

# B. Aggregate Total Word Count
df['total_review_word_count'] = (
    df['benefitsReview_word_count'] + 
    df['sideEffectsReview_word_count'] + 
    df['commentsReview_word_count']
)

# C. Ordinal Numerical Mapping
side_effects_map = {
    'No Side Effects': 0,
    'Mild Side Effects': 1,
    'Moderate Side Effects': 2,
    'Severe Side Effects': 3,
    'Extremely Severe Side Effects': 4
}
df['sideEffect_severity_score'] = df['sideEffects'].map(side_effects_map)

effectiveness_map = {
    'Ineffective': 0,
    'Marginally Effective': 1,
    'Moderately Effective': 2,
    'Considerably Effective': 3,
    'Highly Effective': 4
}
df['effectiveness_score'] = df['effectiveness'].map(effectiveness_map)

# D. Binary Satisfaction Feature
df['is_high_satisfaction'] = (df['rating'] >= 7).astype(int)

print(f"Expanded Dataset Dimensions: {df.shape[0]} rows, {df.shape[1]} columns\n")
print("Engineered Columns List:")
for idx, col in enumerate(df.columns, 1):
    print(f"{idx}. {col}")
```

  <ul style="font-family: sans-serif;">
    <li>Expanded feature set from <strong>8 columns to 18 features</strong> 
    <li>Constructed numerical metrics for word counts, total review length, ordinal scales for effectiveness and side effects, and a satisfaction classification flag.</li>

<h2 style="color:white;">
📍 Outlier Detection and Treatment Using the IQR Method
</h2>

</div>

```python
# Select numerical features for outlier detection

numeric_features = [
    'rating',
    'sideEffect_severity_score',
    'effectiveness_score',
    'total_review_word_count',
    'benefitsReview_word_count',
    'sideEffectsReview_word_count'
]
```

```python
# Detect outliers using the IQR method

outlier_summary = []

for column in numeric_features:
    
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outlier_count = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ).sum()
    
    outlier_summary.append({
        'Feature': column,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': outlier_count
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

  <ul style="font-family: sans-serif;">
    <li>The IQR method detected 146 potential outliers in total review word count, 109 in benefits review word count, and 152 in side-effects review word count, while no outliers were detected in rating, effectiveness score, or side-effect severity score. As these long reviews may represent genuine patient feedback, the records were retained without modification.</li>

```python
rating_pivot = pd.pivot_table(
    df,
    values='rating',
    index='sideEffect_severity_score',
    columns='effectiveness_score',
    aggfunc='mean'
).round(2)

# Rename effectiveness columns
rating_pivot.columns = [
    'Ineffective',
    'Marginally Effective',
    'Moderately Effective',
    'Considerably Effective',
    'Highly Effective'
]

# Rename side-effect rows
rating_pivot.index = [
    'No Side Effects',
    'Mild Side Effects',
    'Moderate Side Effects',
    'Severe Side Effects',
    'Extremely Severe Side Effects'
]

rating_pivot
```

```python
# Create grouped summary

grouped_scores = df.groupby(
    ['effectiveness_score', 'sideEffect_severity_score']
).agg(
    mean_rating=('rating', 'mean'),
    patient_count=('rating', 'count')
).reset_index()

# Convert scores to meaningful labels
effectiveness_labels = {
    0: 'Ineffective',
    1: 'Marginally Effective',
    2: 'Moderately Effective',
    3: 'Considerably Effective',
    4: 'Highly Effective'
}

side_effect_labels = {
    0: 'No Side Effects',
    1: 'Mild Side Effects',
    2: 'Moderate Side Effects',
    3: 'Severe Side Effects',
    4: 'Extremely Severe Side Effects'
}

grouped_scores['effectiveness'] = (
    grouped_scores['effectiveness_score'].map(effectiveness_labels)
)

grouped_scores['side_effects'] = (
    grouped_scores['sideEffect_severity_score'].map(side_effect_labels)
)

# Create visualization
fig2 = px.scatter(
    grouped_scores,
    x='effectiveness',
    y='side_effects',
    size='patient_count',
    color='mean_rating',
    color_continuous_scale='RdYlGn',
    title='Effectiveness and Side Effects vs. Mean Patient Rating',
    labels={
        'effectiveness': 'Medication Effectiveness',
        'side_effects': 'Side-Effect Severity',
        'mean_rating': 'Mean Rating',
        'patient_count': 'Number of Patients'
    }
)

fig2.update_layout(
    font_family='sans-serif',
    height=500
)

fig2.show()
```
